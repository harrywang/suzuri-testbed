# Pilot: gain by bundle size
Simulated paired runs for `notes/study-design.md`.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
modules = np.arange(1, 7)
gain = np.array([0.10, 0.16, 0.18, 0.14, 0.09, 0.05])
bare = rng.binomial(200, 0.42, 6) / 200
skill = rng.binomial(200, 0.42 + gain) / 200

fig, ax = plt.subplots(figsize=(7, 2.8), dpi=300)
width = 0.38
ax.bar(modules - width / 2, bare, width, color="#b8b8b8", label="no skill")
ax.bar(modules + width / 2, skill, width, color="#c0472b", label="curated skill")
ax.set(xlabel="modules in the skill bundle", ylabel="pass rate", ylim=(0, 0.8))
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, ncols=2, loc="upper center");
```
